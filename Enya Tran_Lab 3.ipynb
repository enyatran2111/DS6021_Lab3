### Part 1

**1a.** Model 1 can be written as $\mathbf y=\mathbf X\boldsymbol\beta+\boldsymbol\varepsilon$:

$$
\begin{bmatrix}
\text{price}_1\\
\text{price}_2\\
\vdots\\
\text{price}_{3000}
\end{bmatrix}
=
\begin{bmatrix}
1 & \text{carat}_1 & \text{depth}_1 & \text{table}_1\\
1 & \text{carat}_2 & \text{depth}_2 & \text{table}_2\\
\vdots & \vdots & \vdots & \vdots\\
1 & \text{carat}_{3000} & \text{depth}_{3000} & \text{table}_{3000}
\end{bmatrix}
\begin{bmatrix}
\beta_0\\
\beta_1\\
\beta_2\\
\beta_3
\end{bmatrix}
+
\begin{bmatrix}
\varepsilon_1\\
\varepsilon_2\\
\vdots\\
\varepsilon_{3000}
\end{bmatrix}.
$$

Here, $\mathbf y$ contains the 3,000 prices. Each row of $\mathbf X$ represents one diamond, and the first column of 1s allows for an intercept. So $\mathbf y$ and $\boldsymbol\varepsilon$ are $3000\times1$, $\mathbf X$ is $3000\times4$, and $\boldsymbol\beta=[\beta_0,\beta_1,\beta_2,\beta_3]^\top$ is $4\times1$.

**1b.** We want the coefficients that make the sum of squared errors as small as possible, so we differentiate SSE with respect to the coefficient vector and set the result equal to zero. This gives $\mathbf X^\top\mathbf X\hat{\boldsymbol\beta}=\mathbf X^\top\mathbf y$. Because the columns of $\mathbf X$ are linearly independent, $\mathbf X^\top\mathbf X$ has an inverse, giving $\hat{\boldsymbol\beta}=(\mathbf X^\top\mathbf X)^{-1}\mathbf X^\top\mathbf y$.

**1c.** The carat coefficient in Model 1 is $7,874.15, so a diamond that is 0.10 carats heavier has a predicted price that is $0.10\times \$7{,}874.15=\$787.41$ higher when depth and table are the same. That is a model prediction, not a guaranteed difference in actual prices, since other factors also affect price.

### Part 2

**2a.** Dropping one reference category leaves 4 cut indicators, 6 color indicators, and 7 clarity indicators: 17 in tota. If we kept every indicator for any one of those variables, its columns would add up to 1 for every diamond. That would duplicate the intercept column and create exact multicollinearity.

**2b.** The cut_Ideal coefficient is about $930.29. This means Model 2 predicts an Ideal diamond to cost $930.29 more than a Fair diamond with the same carat, depth, table, color, and clarity. Fair is the reference cut category.

**2c.** The raw mean price of Ideal diamonds is $279.38 lower than that of Fair diamonds, but the cut_Ideal coefficient is +$930.29. The raw comparison does not account for differences in other characteristics; for example, Ideal diamonds in this sample are smaller on average. The coefficient holds the other included predictors constant, but it still shows an association, not that changing a diamond’s cut would cause its price to rise by $930.29.

### Part 3

**3a.** The bootstrap standard error for carat is about $89.26. It is the standard deviation of the 1,000 carat coefficients obtained by resampling and refitting the model, so it shows how much the estimate varies across those samples. We resample each diamond’s predictors and price together to preserve the relationship between its characteristics and its price.

**3b.** The 95% bootstrap interval is [$8,715.53, $9,063.45] for carat and [$634.63, $1,222.12] for cut_Ideal. Neither interval contains zero, which supports a positive conditional association with price for each predictor. These are ranges for plausible *coefficient values*, not ranges of prices for an individual diamond; an individual prediction also has unexplained price variation.

**3c.** Raising B from 1,000 to 10,000 would make the bootstrap interval less sensitive to which resamples happened to be drawn. It would still be resampling the same 3,000 diamonds, so it would not add new information or necessarily make the interval narrower. Collecting more independent diamonds could reduce the uncertainty in the coefficient estimates.

### Part 4

**4a.** A residual of +$1,200 means the actual price was $1,200 above the predicted price, so the model underpredicted it. A residual of −$1,200 means the actual price was $1,200 below the prediction, so the model overpredicted it.

**4b.** In the residuals-versus-fitted-price plot, the residuals tend to be above zero at low fitted prices, below zero in the middle, and above zero again at higher fitted prices. That curved pattern suggests the model is missing a nonlinear relationship. In the residuals-versus-carat plot, the points spread out more as carat increases, especially beyond about one carat, suggesting the error variance is not constant.

**4c.** I would try adding a carat-squared term so the model can capture a curved relationship between carat and price. A high $R^2$ can coexist with a clear pattern in the residuals, and narrow coefficient intervals do not mean the model has captured every pattern in the data.

### Part 5

**5a.** Model 2’s $R^2$ is 0.9195, so it explains about 91.95% of the variation in price among the 3,000 diamonds used to fit it**. This does not mean it predicts 91.95% of individual diamond prices correctly; $R^2$ measures overall fit to this sample.

**5b.** Model 1 has $p=3$ and adjusted $R^2=0.8569$. Model 2 has $p=20$ and adjusted $R^2=0.9190$. Counting only six original variables for Model 2 would miss the 17 indicator columns created from cut, color, and clarity; together with the 3 numerical columns, that gives 20 fitted predictors. Model 2 has the stronger adjusted fit.

**5c.** Ordinary $R^2$ cannot go down here because Model 2 includes everything in Model 1: if the new indicators do not help, OLS can give them coefficients of zero and keep the same fit. Adjusted $R^2$ can go down because it accounts for the added predictor columns. Neither statistic tells us how the model will perform on new diamonds, since both were calculated using the diamonds that fitted the models.
